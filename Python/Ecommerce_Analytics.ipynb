# 5. Feature Engineering

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df["Order Year"] = df["Order Date"].dt.year
```

```python
df["Order Month"] =df["Order Date"].dt.month
```

```python
df.columns
```

```python
df["Processing Time"] = df["Ship Date"] - df["Order Date"]
```

```python
df["Profit_Margin"] = df["Profit"] / df["Sales"]
```

# 6. Exploratory Data Analysis (EDA)

## 6.1 Sales Distribution (Histogram)

```python
plt.figure(figsize = (10,5))

sns.histplot(df["Sales"], bins=30, color = "pink")

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Number of Orders")

plt.show()
```

Histogram - Bars touch each other 
Why? - Because Sales values are continuous

## 6.2 Sales Outlier Detection (Box Plot)

```python
plt.figure(figsize=(10,4))
sns.boxplot(x=df["Sales"])        
plt.title("Box Plot of Sales")
plt.show()
```

Blue Box - Starts around 25th percentile (Q1) & ends around 75th percentile(Q3)
According to the BLUE BOX - Half of all our customer orders are inside blue box 

Median - Dark vertical Line (Middle order)
Tells us: 50% orders are smaller than this value & 50% orders are larger than this value
#median hardly changes

Whiskers - Lines
These are still part of the normal range, extend from the box to the Largest & Smallest values that are still considered normal

Outliers - Dots
Each Dot represents 1 order whose sales value is unusually high, compared to most other orders.
There are Unusually Large sales, No unusual small sales.

## 6.4 Total Sales by Category

**Which Category generates the highest sales?**

```python
category_sales = df.groupby("Category")["Sales"].sum()
```

```python
category_sales
```

groupby() - Return unique values by grouping them into one

```python
df["Category"].unique()
```

**BAR CHART**

```python
plt.figure(figsize=(8,5))

category_sales.plot(kind="bar", color = ["pink"])

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

First I grouped data using groupby() and calculated total sales using sum().
Then, I used a bar chart because it is most suuitable gor visualization for comparing values across categories.

## 6.5 Total Sales by Sub-Category

```python
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum()
```

```python
subcategory_sales
```

```python
len(subcategory_sales)
```

```python
subcategory_sales = subcategory_sales.sort_values(ascending=False)
subcategory_sales
```

sort_values(ascending=False)
New Pandas method - sort_values = function, ascending=False = Parameter

```python
#Creating Bar Chart for Sub-Category Sales
plt.figure(figsize=(12,6))

subcategory_sales.plot(kind="bar", color="pink")

plt.title("Total Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales")

plt.show()
```

## 6.6 Total Profit by Sub_Category

```python
subcategory_profit = df.groupby("Sub-Category")["Profit"].sum()
subcategory_profit
```

```python
subcategory_profit = subcategory_profit.sort_values(ascending=False)
subcategory_profit
```

```python
#Creating Bar Chart for Total Profit by Sub-Category
plt.figure(figsize=(12,6))

subcategory_profit.plot(kind="bar", color="pink")

plt.title("Total Profit by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Profit")

plt.show()
```

## 6.7 Total Sales by Region

```python
region_sales = df.groupby("Region")["Sales"].sum()

region_sales = region_sales.sort_values(ascending=False)

plt.figure(figsize=(8,5))
region_sales.plot(kind="bar", color="orange")

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.show()
```

## 6.8 Total Sales by Segment

## 6.9 Total Profit by Region

## 6.11 Monthly Sales Trend

**LINE CHART**

How did sales change over time?

```python
df["Month-Year"] = df["Order Date"].dt.to_period("M")
```

```python
monthly_sales = df.groupby("Month-Year")["Sales"].sum()
```

```python
plt.figure(figsize=(12,6))

monthly_sales.plot(kind="line", marker="o", color="blue")

plt.title("Monthly Sales Trend")
plt.xlabel("Month-Year")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.show()
```

## 6.12 Sales vs Profit Relationship (Scatter Plot)

**SCATTER PLOT**

We use scatter plot to know, If there is any relationship between Sales & Profit?
1 Dot = 1 Order (1 Transaction) Every single Dot tells the story of 1 Order

```python
plt.figure(figsize=(8,6))

sns.scatterplot(x=df["Sales"], y=df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

## 6.13 Correlation Heatmap

**Correlation Heatmap**

Which numeric columns are related to each other?

```python
numeric_df = df.select_dtypes(include="number")

correlation = numeric_df.corr()

plt.figure(figsize=(8,6))

sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

# 7. Final Business Insights

1. Most customer orders are of low sales value, while only a few orders have very high sales.

2. Technology generated the highest total sales among all categories.

3. Phones generated the highest total sales among all sub-categories.

4. Copiers generated the highest total profit.

5. Tables, Bookcases, and Supplies showed negative overall profit.

6. The West region generated the highest sales.

7. The Consumer segment generated the highest sales.

8. Monthly sales fluctuated over time, with stronger sales towards the later months.

9. High sales do not always result in high profit.

10. Higher discounts tend to reduce profit.
